```python
import pandas as pd

# 1. Đọc dữ liệu
df_reviews = pd.read_csv('/content/reviews.csv')
df_items = pd.read_csv('/content/order_items.csv', low_memory=False)

# 2. Merge dữ liệu để xác thực FK
df_review_combined = pd.merge(
    df_reviews,
    df_items[['order_id', 'product_id']],
    on=['order_id', 'product_id'],
    how='inner'
)

# 3. Chọn lọc và đổi tên cột theo yêu cầu
# Cấu trúc: review_id, order_id_FK, product_id_FK, review_date, rating, review_title
review_final = df_review_combined[[
    'review_id',
    'order_id',
    'product_id',
    'review_date',
    'rating',
    'review_title'
]].copy()

review_final.columns = [
    'review_id',
    'order_id_FK',
    'product_id_FK',
    'review_date',
    'rating',
    'review_title'
]

# 4. Lưu file
output_review_path = '/content/review_new.csv'
review_final.to_csv(output_review_path, index=False)

print(f'Đã tạo file thành công tại: {output_review_path}')
display(review_final.head())
```
