```python
import pandas as pd

# 1. Đọc dữ liệu nguồn
df_web = pd.read_csv('/content/web_traffic.csv')

# 2. Lọc và đổi tên các cột theo cấu trúc yêu cầu
# Cấu trúc yêu cầu: date, traffic_source, sessions, unique_visitors, page_views, bounce_rate, avg_session_duration
web_traffic_final = df_web[[
    'date',
    'traffic_source',
    'sessions',
    'unique_visitors',
    'page_views',
    'bounce_rate',
    'avg_session_duration_sec'
]].copy()

web_traffic_final.columns = [
    'date',
    'traffic_source',
    'sessions',
    'unique_visitors',
    'page_views',
    'bounce_rate',
    'avg_session_duration'
]

# 3. Lưu ra file mới
output_web_path = '/content/web_traffic_new.csv'
web_traffic_final.to_csv(output_web_path, index=False)

print(f'Đã tạo file thành công tại: {output_web_path}')
display(web_traffic_final.head())
```
