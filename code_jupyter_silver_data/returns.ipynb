```python
import pandas as pd
from pathlib import Path

# 1. Đọc dữ liệu nguồn
df_returns = pd.read_csv('/content/returns.csv')
df_items = pd.read_csv('/content/order_items.csv')

# 2. Thực hiện merge (join) để đảm bảo tính nhất quán của dữ liệu
# Mặc dù returns.csv thường đã có đủ thông tin, việc join với order_items giúp xác thực các FK
returns_new = pd.merge(
    df_returns,
    df_items[['order_id', 'product_id']],
    on=['order_id', 'product_id'],
    how='inner'
)

# 3. Lọc và đổi tên các cột theo cấu trúc yêu cầu
# Cấu trúc: return_id, order_id_FK, product_id_FK, return_date, return_reason, return_quantity, refund_amount
returns_new = returns_new[[
    'return_id',
    'order_id',
    'product_id',
    'return_date',
    'return_reason',
    'return_quantity',
    'refund_amount'
]]

returns_new.columns = [
    'return_id',
    'order_id_FK',
    'product_id_FK',
    'return_date',
    'return_reason',
    'return_quantity',
    'refund_amount'
]

# 4. Lưu ra file mới
output_path = '/content/returns_new.csv'
returns_new.to_csv(output_path, index=False)

print(f'Đã tạo file thành công tại: {output_path}')
display(returns_new.head())
```
