```python
import pandas as pd

# Đọc thử dữ liệu từ các file liên quan để phân tích cấu trúc
customers_df = pd.read_csv('/content/customers_new.csv')
geography_df = pd.read_csv('/content/geography_new.csv')

print("Customers columns:", customers_df.columns)
print("Geography columns:", geography_df.columns)

display(customers_df.head(2))
display(geography_df.head(2))
```

```python
# Tiến hành xử lý và mapping dữ liệu theo schema DIM_CUSTOMER
# Thực hiện merge bảng customers_new (zip_FK) với geography_new (zip)
merged_df = pd.merge(
    customers_df,
    geography_df,
    left_on='zip_FK',
    right_on='zip',
    how='left'
)

# Tạo khóa chính customer_key kiểu số nguyên tự tăng từ 1
merged_df = merged_df.reset_index(drop=True)
merged_df['customer_key'] = merged_df.index + 1

# Khởi tạo từ điển ánh xạ / định nghĩa cột theo schema
schema_columns = {
    'customer_key': 'customer_key',
    'customer_id': 'customer_id',
    'signup_date': 'signup_date',
    'gender': 'gender',
    'age_group': 'age_group',
    'acq_channel': 'acq_channel',
    'city': 'city',
    'region': 'region',
    'district': 'district',
    'zip': 'zip'
}

# Đảm bảo tất cả các cột của schema đều có mặt
for col in schema_columns.keys():
    if col not in merged_df.columns:
        merged_df[col] = None

dim_customers = merged_df[list(schema_columns.keys())].copy()

# Ghi dữ liệu ra file mới không kèm index tự động của pandas
dim_customers.to_csv('/content/dim_customers.csv', index=False)

print("Đã sửa và tạo file /content/dim_customers.csv thành công!")
display(dim_customers.head())
```

```python
import pandas as pd

# Đọc dữ liệu địa lý
geography_df = pd.read_csv('/content/geography_new.csv')

# Loại bỏ các dòng trùng lặp nếu có để đảm bảo tính duy nhất
dim_geo_df = geography_df.drop_duplicates(subset=['zip']).copy()

# Tạo khóa chính tự tăng geography_key bắt đầu từ 1
dim_geo_df = dim_geo_df.reset_index(drop=True)
dim_geo_df['geography_key'] = dim_geo_df.index + 1

# Sắp xếp và lựa chọn các cột theo đúng schema DIM_GEOGRAPHY
schema_columns = ['geography_key', 'zip', 'city', 'region', 'district']
dim_geography = dim_geo_df[schema_columns]

# Xuất file kết quả
dim_geography.to_csv('/content/dim_geography.csv', index=False)

print("Đã tạo file /content/dim_geography.csv thành công!")
display(dim_geography.head())
```

```python
import pandas as pd

# Xác định khoảng ngày bắt đầu và kết thúc từ các nguồn dữ liệu có sẵn
# Quét cột signup_date từ customers_df
customers_df['signup_date'] = pd.to_datetime(customers_df['signup_date'], errors='coerce')
min_date = customers_df['signup_date'].min()
max_date = customers_df['signup_date'].max()

# Đọc thêm từ bảng orders (nếu có) để bao phủ đầy đủ ngày trong hệ thống
try:
    orders_df = pd.read_csv('/content/orders_new.csv')
    orders_df['order_date'] = pd.to_datetime(orders_df['order_date'], errors='coerce')
    min_date = min(min_date, orders_df['order_date'].min())
    max_date = max(max_date, orders_df['order_date'].max())
except Exception:
    pass

# Nếu có bất kỳ giá trị NaT nào, gán khoảng mặc định dự phòng hợp lý
if pd.isnull(min_date): min_date = pd.to_datetime('2010-01-01')
if pd.isnull(max_date): max_date = pd.to_datetime('2030-12-31')

print(f"Date Range: {min_date.date()} to {max_date.date()}")

# Tạo chuỗi ngày liên tục
date_range = pd.date_range(start=min_date, end=max_date, freq='D')
dim_date_df = pd.DataFrame({'full_date': date_range})

# Tạo khóa chính tự tăng bắt đầu từ 1
dim_date_df['date_key'] = dim_date_df.index + 1

# Tính toán các thuộc tính lịch chi tiết theo Schema
dim_date_df['full_date'] = dim_date_df['full_date'].dt.date
dim_date_df['day_of_week'] = pd.to_datetime(dim_date_df['full_date']).dt.dayofweek + 1 # 1: Thứ 2, ..., 7: Chủ Nhật
dim_date_df['day_name'] = pd.to_datetime(dim_date_df['full_date']).dt.day_name()
dim_date_df['day_of_month'] = pd.to_datetime(dim_date_df['full_date']).dt.day
dim_date_df['month'] = pd.to_datetime(dim_date_df['full_date']).dt.month
dim_date_df['month_name'] = pd.to_datetime(dim_date_df['full_date']).dt.month_name()
dim_date_df['quarter'] = pd.to_datetime(dim_date_df['full_date']).dt.quarter
dim_date_df['year'] = pd.to_datetime(dim_date_df['full_date']).dt.year
dim_date_df['is_weekend'] = dim_date_df['day_of_week'].isin([6, 7])

# Chọn lọc và sắp xếp cột theo cấu trúc Schema yêu cầu
schema_columns = [
    'date_key', 'full_date', 'day_of_week', 'day_name', 'day_of_month',
    'month', 'month_name', 'quarter', 'year', 'is_weekend'
]
dim_date = dim_date_df[schema_columns].copy()

# Xuất file kết quả
dim_date.to_csv('/content/dim_date.csv', index=False)

print("Đã tạo file /content/dim_date.csv thành công!")
display(dim_date.head())
```
