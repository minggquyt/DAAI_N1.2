```python
import pandas as pd

# --- 1. Tải dữ liệu từ file products.csv ---
# Đọc file CSV vào DataFrame để bắt đầu tiền xử lý
products_df = pd.read_csv('/content/products.csv')

print("Dữ liệu gốc (5 dòng đầu):")
display(products_df.head())

# --- 2. Kiểm tra trùng lặp Khóa chính (product_id) ---
# Dựa trên schema, product_id là Primary Key nên không được phép trùng lặp
duplicate_ids = products_df[products_df.duplicated(subset=['product_id'], keep=False)]

if not duplicate_ids.empty:
    print(f"Phát hiện {len(duplicate_ids)} dòng bị trùng lặp product_id. Tiến hành xóa và giữ lại dòng đầu tiên:")
    products_df.drop_duplicates(subset=['product_id'], keep='first', inplace=True)
else:
    print("\nKhông tìm thấy trùng lặp cho khóa chính product_id.")

# --- 3. Tiền xử lý dữ liệu Null thành dữ liệu mặc định ---
# Điền các giá trị thiếu dựa trên kiểu dữ liệu của schema

# product_id: VARCHAR - Điền 'UNKNOWN_ID' nếu null
products_df['product_id'] = products_df['product_id'].fillna('UNKNOWN_ID').astype(str)

# product_name, category, segment, size, color: VARCHAR - Điền 'N/A'
string_cols = ['product_name', 'category', 'segment', 'size', 'color']
for col in string_cols:
    if col in products_df.columns:
        products_df[col] = products_df[col].fillna('N/A').astype(str)

# price, cogs: DECIMAL (float) - Điền 0.0
float_cols = ['price', 'cogs']
for col in float_cols:
    if col in products_df.columns:
        products_df[col] = products_df[col].fillna(0.0).astype(float)

# --- 4. Lọc cột theo Schema mong muốn ---
# Đảm bảo các cột tuân thủ đúng cấu trúc lớp PRODUCT
expected_columns = ['product_id', 'product_name', 'category', 'segment', 'size', 'color', 'price', 'cogs']
processed_products_df = products_df[expected_columns]

print("\nThông tin dữ liệu sau khi xử lý:")
processed_products_df.info()

# --- 5. Tạo file mới newproduct.csv ---
output_path = '/content/newproduct.csv'
processed_products_df.to_csv(output_path, index=False)

print(f"\nDữ liệu đã được xử lý và lưu thành công tại: {output_path}")
```
